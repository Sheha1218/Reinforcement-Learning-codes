```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([-2,-1,0,1,2],dtype=np.float32)
y=np.array([-2,-1,0,1,2],dtype=np.float32)
```

```python
def forward(w,x):
    return w*x
```

```python
w=0.5

y_predicted=forward(w,x)
print(y_predicted)

plt.figure()
ax=plt.subplot()
ax.scatter(x,y)
ax.set_xlabel('X')
ax.set_ylabel('Y')

ax.axhline(color="lightgray")
ax.axvline(color="lightgray")
```
